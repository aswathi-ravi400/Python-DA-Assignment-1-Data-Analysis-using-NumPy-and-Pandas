Numpy Array Operations:
Problem Statement: To learn how to use NumPy for creating arrays, performing basic

1. Create a 1D NumPy Array:
Create a 1D NumPy array named temperatures_w1 for Week 1 with the

[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9]

```python
import numpy as np

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print(temperatures_w1)
```

```python
print("Mininum Temperature:",temperatures_w1.min())
print("Maximum Temperature:",temperatures_w1.max())
print("Mean Temperatures of the week:",temperatures_w1.mean())
```

Pandas Series:
1. Creating Pandas Series:
● Create a Pandas Series named marks with the following data:
➢ Marks: 95, 92, 89, 85, 80
➢ Custom Indices: 'Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'

Ans:Installed pandas.A Pandas Series named marks was created using the given marks.Custom indexes from Rank1 to Rank5 were assigned to identify each student's rank.

```python
%pip install pandas
```

```python
import pandas as pd
marks=pd.Series(
    [95,92,89,85,80],
    index=['Rank1','Rank2','Rank3','Rank4','Rank5']
)
print(marks)
```

2. Indexing and Slicing:
● Use the integer index position to access the mark of the 1st rank student.
● Use the loc accessor to retrieve the marks of the top 3 ranks by
specifying their index labels.
● Use the iloc accessor to retrieve the mark of the 3rd rank student.
● Apply a boolean mask to filter the Series and retrieve the ranks where the
marks are greater than 90.


Ans:Integer index position:Used to access the mark of the first-rank student.
loc: Used to retrieve the marks of the top three ranks using their index labels.
iloc: Used to retrieve the mark of the third-rank student using its position.
Boolean masking: Used to filter the Series and display ranks with marks greater than 90.

```python
print("1st Rank Student:",marks.iloc[0])
print("Top 3 Ranks:",marks.loc[['Rank1','Rank2','Rank3']])
print("Marks of 3rd Rank Student:",marks.iloc[2])
print("Marks greater than 90:",marks[marks>90])

```

3. Manipulating Series:
● Modify the mark of the student with the 1st rank to 100.
● Remove the entry corresponding to the last rank from the Series.
● Compute the CGPA by dividing each mark in the Series by 10.

Ans:The mark of Rank1 was changed to 100.The entry corresponding to Rank5 was removed.The CGPA was calculated by dividing each remaining mark by 10.

```python
marks.loc['Rank1']=100
print(marks)
```

```python
marks=marks.drop('Rank5')
print(marks)
```

```python
CGPA=marks/10
print(CGPA)
```

Pandas DataFrame:
1. Creating Pandas DataFrame:
● Create a Pandas DataFrame named transactions with the following data:
➢ TransactionID: 101, 102, 103, 104, 105, 106, 107, 108, 109, 110
➢ ProductCategory: 'Electronics', 'Clothing', 'Electronics', 'Furniture',
'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'
➢ Region: 'North', 'South', 'North', 'East', 'West', 'North', 'East', 'West',
'South', 'North'
➢ Amount: 200, 150, 300, 450, 200, 250, 300, 180, 350, 400

Ans:A DataFrame named transactions was created using transaction IDs,product categories,regions and transaction amounts.

```python
import pandas as pd
transactions=pd.DataFrame({
    'TransactionID':[101,102,103,104,105,106,107,108,109,110],
    'ProductCategory':['Electronics','Clothing','Electronics', 'Furniture',
                       'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
    'Region':['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West',
              'South', 'North'],
    'Amount':[200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})
print(transactions)
```

3. Manipulating the DataFrame:
● Modify the 'Amount' for the transaction with TransactionID 102 to 165.

Ans:The amount for TransactionID 102 was changed to 165 using the loc accessor.

```python
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
print(transactions[transactions['TransactionID'] == 102])
```

● Add a new column 'Discount' by calculating 10% of the 'Amount'

Ans:A new column named Discount was added by calculating 10% of each transaction's amount.

```python
transactions['Discount'] = transactions['Amount'] * 0.10
print(transactions)
```

● Remove the row with TransactionID 109.

Ans:The row with TransactionID 109 was removed from the DataFrame by filtering out that transaction.

```python
transactions = transactions[transactions['TransactionID'] != 109]
print(transactions)
```

● Delete the 'Discount' column from the DataFrame.

Ans:The Discount column was deleted using the drop()function,leaving the other columns unchanged.

```python
transactions = transactions.drop(columns=['Discount'])
print(transactions)
```
